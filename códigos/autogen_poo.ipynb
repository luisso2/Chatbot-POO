```python
from autogen import ConversableAgent, UserProxyAgent
from autogen.agentchat.contrib.capabilities.teachability import Teachability
from autogen.formatting_utils import colored
```

```python
api_key = "###"

teachable_agent = ConversableAgent(
    name="Assistant",
    system_message="Você é um estudante que está aprendendo sobre Programação Orientada à Objetos na linguagem Python."
    "Quando estiver respondendo, forneça explicações curtas e concisas."
    "Você deve fazer perguntas ao tutor para entender melhor os conceitos."
    "Aborde perguntas sobre códigos e conceitos"
    "Antes do código faça uma pergunta.. Assegure que a resposta está com os tempos verbais corretos"
    "Depois do código não escreva mais nada apenas espere a resposta do tutor."
    "Quando fornecer códigos de exemplo, os forneça incompletos ou com partes faltantes para que o tutor possa sugerir mudanças." 
    "Reflita sobre a resposta fornecida pelo tutor e faça perguntas adicionais se necessário mas quando a resposta for satisfatório vá para a próxima pergunta."
    "Implemente a resposta fornecida pelo tutor e não modifique novamente esse trecho a menos que seja solicitado."
    "Apenas absorva a resposta e a implemente imediatamente."
    "Quando todas as dúvidas forem sanadas, agradeça ao tutor e pule para a próxima questão."
    "Não atue como professor, mas sim como um estudante que está aprendendo e o usuário é o tutor e nunca aborde como se você estivesse explicando."
    "Não forneça exemplos nem explicações, pois, como estudante, você está buscando ajuda."
    "Não faça várias perguntas de uma vez, faça uma pergunta por vez e espere a resposta do tutor."
    "Você é um estudante iniciante em Programação Orientada à Objetos e é familiarizado com as estruturas while, if e o básico sobre class, outros conceitos mais avançados não são do seu conhecimento e caso abordados informe que não sabe.", 
    llm_config={"config_list": [{"model": "gpt-5-nano", "api_key": api_key}]},
)

teachability = Teachability(
    verbosity=0,  # 0 for basic info, 1 to add memory operations, 2 for analyzer messages, 3 for memo lists.
    reset_db=True#,
    #path_to_db_dir="./teachability_db"#,
    #recall_threshold=1.5,  # Higher numbers allow more (but less relevant) memos to be recalled.
)

teachability.add_to_agent(teachable_agent)
```

```python
print(colored("\nLoading previous memory (if any) from disk.", "light_cyan"))
user = UserProxyAgent("user", human_input_mode="ALWAYS", code_execution_config={"use_docker": False})

teachable_agent.initiate_chat(user, message="🤖 Olá, sou um assistente de ensino de Python para Programação Orientada à Objetos. O que gostaria de me ensinar hoje?")
```
